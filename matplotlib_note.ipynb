```python
#Format Strings fmt------------------------
'''You can also use the shortcut string notation parameter to specify the marker.
This parameter is also called fmt, and is written with this syntax:
marker|line|color
'''
#example
import matplotlib.pyplot as plt
import numpy as np
ypoints = np.array([3, 8, 1, 10])
plt.plot(ypoints, 'o:r')
plt.show()

#example without shortcut
ypoints2=np.array([3,7,6,0,2,9,3])
xpoint2=np.array([2,5,9,1,4,3,6])
plt.plot(ypoints2,xpoint2,marker='*', ms=20, mec='r', mfc='r', linestyle='--',linewidth='10.4')

#Line Reference
'''
Line Syntax	| Description
'-'	          Solid line	
':'	          Dotted line	
'--'	      Dashed line	
'-.'	      Dashed/dotted line'''

#Color Reference
'''
Color Syntax	Description
'r'          	Red	
'g'	            Green	
'b'	            Blue	
'c'	            Cyan	
'm'	            Magenta	
'y'	            Yellow	
'k'         	Black	
'w'         	White'''

#Marker Size----------------------------------------------------
'''you can rsize your marker by providing "ms=" argument'''
#example
# ypoints3 = np.array([3, 8, 1, 10])
# plt.plot(ypoints3, marker = 'o', ms = 20)
# plt.show()
```

```python
#-------------------Matplotlib Labels and Title----------------------
'''With Pyplot, you can use the xlabel() and ylabel() functions to set a label
for the x- and y-axis.
and title by using 'title()'
'''
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)

plt.title("sports watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.show()
```

```python
#----------------Display Multiple Plots----------------------
'''With the subplot() function you can draw multiple plots in one figure:'''

#example
import matplotlib.pyplot as plt
import numpy as np
#plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])
plt.subplot(1, 2, 1)
plt.plot(x,y)
#plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 2)

plt.plot(x,y)
plt.show()

#example

#fig,ax=plt.subplots(nrows,ncols,figsize=(width, height))
fig, ax=plt.subplots(1,2, figsize=(10,5))
x=[1,2,3,4]
y=[10,20,15,25]
ax[0].plot(x,y, color='red')
ax[0].set_title('Line plot')

ax[1].bar(x,y, color='green')
ax[1].set_title('Bar chart')

fig.suptitle("comparison of line and bar charts")
# fig, ax=plt.subplots(2,1,sharex=True)
plt.tight_layout()#this use for fit/fix the size of graph
plt.show()
```

```python
#Revision by creating small graph:- 
#question:-
'''
            create a graph of month sales car production month given [1,2,3,4,5], and each month saled have [1000, 5000, 4000, 90000, 8500] show in the graph
'''
```

```python
#answer
import numpy as np
import matplotlib.pyplot as plt

months=np.array([1,2,3,4,5])
sales=np.array([1000, 5000, 4000, 9000, 5500])
plt.plot(months,sales, color='green', linestyle='--', linewidth='1', marker='o',label='2026 sales data')
plt.title('Car sales')
plt.xlabel('months')
plt.ylabel('number of cars sale')
plt.grid(color='gray', linestyle=':', linewidth='0.4')
plt.legend(loc='upper left', fontsize=12)
plt.xlim(1,7)
plt.ylim(1000,10000)
plt.xticks([1,2,3,4,5,6,7],['M1','M2','M3','M4','M5','M6','M7',])
plt.show()
```
